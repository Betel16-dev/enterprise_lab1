```python
import pandas as pd

df = pd.read_csv("loan_applications.csv")
print("Dataset shape:")
print(df.shape)

print("\nColumn data types:")
print(df.dtypes)

print("\nFirst five rows:")
print(df.head())
```

```python
# Count missing values in each column
print("Missing values:")
print(df.isnull().sum())

# Count exact duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Remove exact duplicate rows
df_clean = df.drop_duplicates()

# Display cleaned dataset shape
print("Cleaned dataset shape:", df_clean.shape)
```

```python
# Calculate overall probability of default
p_default = (df["defaulted"] == "Yes").mean()

print("P(Default):", p_default)

# Calculate default probability based on collateral
rate_by_collateral = df.groupby("has_collateral")["defaulted"].apply(
    lambda x: (x == "Yes").mean()
)

print(rate_by_collateral)
```

p_default = (df["defaulted"] == "Yes").mean()- df["defaulted"] == "Yes" checks which applicants defaulted.
- True represents Yes, and False represents No.
- .mean() calculates the proportion of Yes values.
Formula:
\[
P(Default)=\frac{\text{Number of defaulted applicants}}{\text{Total applicants with known outcome}}
\]

2. Probability of default with collateral
\[
P(Default\mid Has\ Collateral)
\]

This calculates the percentage of applicants who defaulted among those who offered collateral.

3. Probability of default without collateral
\[
P(Default\mid No\ Collateral)
\]

This calculates the percentage of applicants who defaulted among those who did not offer collateral.
Written answers
You will report three results:
1. Overall probability: P(Default).
2. Probability of default given collateral: P(Default | Has Collateral).
3. Probability of default given no collateral: P(Default | No Collateral).
To calculate how much the rates differ:
\[
Difference = P(D\mid No\ Collateral)-P(D\mid Has\ Collateral)
\]

If the result is positive, applicants without collateral default more often.

For example, if the rates were 30% and 20%, the difference would be 10 percentage points.
