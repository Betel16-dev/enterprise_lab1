```python
import pandas as pd
import os

# Load the dataset
df = pd.read_csv("loan_applications.csv")

# Display the number of rows and columns
print("Dataset shape:", df.shape)

# Display column data types
print("\nData types:")
print(df.dtypes)

# Display the first five rows
print("\nFirst five rows:")
print(df.head())
```

```python
# Calculate mean
print("Mean:", df["loan_amount_etb"].mean())

# Calculate median
print("Median:", df["loan_amount_etb"].median())

# Calculate population standard deviation
print("Population Standard Deviation:",
      df["loan_amount_etb"].std(ddof=0))

# Calculate mode
print("Mode of Credit History:")
print(df["credit_history_flag"].mode())
```

The credit_history_flag column contains categorical values such as Good, Fair, and Poor. The mode identifies the most frequently occurring category. For loan amounts, which are numerical values, mean and median are generally more useful measures of central tendency.

```python
# Check missing values in each column
print("Missing values:")
print(df.isnull().sum())

# Count exact duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Remove exact duplicate rows
df_clean = df.drop_duplicates()

# Display the new dataset shape
print("\nShape after removing duplicates:")
print(df_clean.shape)
```

- isnull().sum() counts missing values in every column.
- duplicated().sum() counts rows that are exact duplicates of earlier rows.
- drop_duplicates() removes exact duplicate rows.
- df_clean.shape displays the dimensions after removing duplicates.


For the combined data-quality problem rate

```python
# Identify rows with missing values
missing_rows = df.isnull().any(axis=1)

# Identify duplicate rows
duplicate_rows = df.duplicated()

# Count rows affected by either problem
problem_rows = (missing_rows | duplicate_rows).sum()

# Calculate percentage
problem_rate = (problem_rows / len(df)) * 100

print("Rows with data quality problems:", problem_rows)
print("Combined problem rate:", problem_rate, "%")
```

| means OR. Therefore, a row is counted if it has at least one missing value, is an exact duplicate, or both. This avoids counting the same row twice.

```python
# Default rates by credit history
print(df.groupby("credit_history_flag")["defaulted"].apply(
    lambda x: (x == "Yes").mean()
))

# Default rates by employment type
print(df.groupby("employment_type")["defaulted"].apply(
    lambda x: (x == "Yes").mean()
))
```

```python
# Calculate overall probability of default
p_default = (df["defaulted"] == "Yes").mean()

print("P(Default):", p_default)

# Calculate default probability based on collateral
rate_by_collateral = df.groupby("has_collateral")["defaulted"].apply(
    lambda x: (x == "Yes").mean()
)

print("\nDefault rate by collateral:")
print(rate_by_collateral)
```
